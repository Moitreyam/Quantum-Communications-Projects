# Superdense Coding

In Quantum Teleportation, we used the concept of Entanglement to transport (or teleport) one qubit using two classical bits. The basis was that both the sender and the receiver must share the same entangled state (Bell state).

In Superdense Coding however, we send two classical bits through one qubit, again using the concept of entanglement. We make use of the fact that a qubit alongwith the shared entangled pair can carry two classical bits worth of information with them.

There are four Bell states which can be distinguishable from one another. So when Bob and Alice both share the same Bell state, and Alice encodes the two classical bits on her part of the shared Bell state, she in return creates one of the four orthogonal Bell states. Upon doing a Bell measurement by Bob on his side, he can retrieve the two classical bits.

In case of a third party person (like an eavesdropper) intercepting Alice's qubit, they will need Bob's qubit in order to make sense of the intercepted entangled state. Without it, they will simply have an entangled state, with nothing to work with. Hence, Superdense Coding is also somewhat secured communication protocol between the sender and the receiver.

We make use of a Bell state say $B_{00}=\frac{1}{\sqrt2}(|00\rangle_{AB} + |11\rangle_{AB})$ which is shared by Alice and Bob.

In order for Alice to send any of the two classical bits $(00),(01),(10),(11)$, she does the following-

1. To send $(00)$-
    She simply sends the unoperated Bell state from her end $B_{00}$
2. To send $(01)$-
   She operates on her share of the Bell state with an X gate to cause a bit flip and create $B_{01}=\frac{1}{\sqrt2}(|01\rangle_{AB} + |10\rangle_{AB})$.
3. To send $(10)$-
   She operates on her share of the Bell state with a phase flip gate i.e. Z and create $B_{10}=\frac{1}{\sqrt2}(|00\rangle_{AB} - |11\rangle_{AB})$.
4. To send $(11)$-
   She operates on her share of the Bell state with both a bit and phase flip using X and Z gates respectively, creating $B_{11}=\frac{1}{\sqrt2}(|01\rangle_{AB} - |10\rangle_{AB})$

The beauty of this is that all the four Bell states are distinguishable and hence a Bell measurement will reveal the sent qubit or Bell state from which Bob can successfully retrieve the classical bits through a Bell measurement.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
n=input("Hi Alice! Please type in your 2-bit values:")

circuit=QuantumCircuit(2,2)

#Alice's end
circuit.h(0)
circuit.cx(0,1)  #entanglement between qubits 1 and 2.

if n=='00':
    pass
elif n=='01':
    circuit.x(0)
elif n=='10':
    circuit.z(0)
elif n=='11':
    circuit.x(0)
    circuit.z(0)
else:
    print("Invalid 2-bit input. Please use 2 digits between 0 and 1")


#Bob's measurement

circuit.cx(0,1)
circuit.h(0)

circuit.measure(0,0)
circuit.measure(1,1)

#Circuit

simulator=AerSimulator()
job=simulator.run(circuit, shots=1000)
result=job.result()
counts=result.get_counts()

measured = list(counts.keys())[0][::-1]
print(f"Alice sent: {n}")
print(f"Bob received: {measured}")
print(f"Shots: {list(counts.values())[0]}/1000")
print(circuit.draw())
```

Thus, using the above, we have successfully demonstrated Superdense Coding wherein using the qubit sent by Alice and the shared entangled state between Alice and Bob, she was able to send two classical bits by sending only one qubit from her end.
